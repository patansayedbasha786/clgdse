```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv("mechian_data_week4.csv")
print("ENGINEERING DATASET")
print(df)
plt.figure(figsize=(7,5))
plt.hist(df["tempature"], bins=5, edgecolor="black")
plt.xlabel("tempature")
plt.ylabel("Number of Machines")
plt.title("Distribution of Machine Temperature")
plt.show()
plt.figure(figsize=(7,5))
plt.scatter(df["tempature"],df["vibration"])
plt.xlabel("tempature")
plt.ylabel("vibration")
plt.title("tempature vs Vibration")
plt.show()
status_count = df["status"].value_counts()
plt.figure(figsize=(7,5))
plt.bar(status_count.index,status_count.values)
plt.xlabel("Machine Status")
plt.ylabel("Number of Machines")
plt.title("Machine Status Distribution")
plt.show()
numeric_data = df[["tempature","Pressure","vibration","Operating_Hours"]]
correlation = numeric_data.corr()
plt.figure(figsize=(8,6))
sns.heatmap(correlation,annot=True,cmap="coolwarm",fmt=".2f")
plt.title("Correlation Heatmap of Engineering Parameters")
plt.show()
```
