```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df=pd.read_csv("week5_dataset.csv")
print(df)
print("-----------------------------------------------------------")
print(df.shape)
print("-----------------------------------------------------------")
print(df.columns)
print("-----------------------------------------------------------")
print(df.dtypes)
print("-----------------------------------------------------------")
df.info()
print("-----------------------------------------------------------")
print(df.isnull().sum())
print("-----------------------------------------------------------")
print(df.describe())
print("-----------------------------------------------------------")
print("mean tempature",df["tempature"].mean())
print("mean tempature",df["tempature"].median())
print("mean tempature",df["tempature"].std())
print("min tempature",df["tempature"].min())
print("max tempature",df["tempature"].max())
print("varience tempature",df["tempature"].var())
print("-----------------------------------------------------------")
numeric_columns=["tempature","pressure","vibration","operating-hours"]
print(df[numeric_columns].mean())
print("-----------------------------------------------------------")
print(df[numeric_columns].median())
print("-----------------------------------------------------------")
print(df[numeric_columns].std())
print("-----------------------------------------------------------")
print(df["status"].value_counts())
print("-----------------------------------------------------------")
correlation=df[numeric_columns].corr()
print(correlation)
plt.figure(figsize=(7,5))
plt.hist(df["tempature"],bins=5,edgecolor="black")
plt.xlabel("tempature")
plt.ylabel("no.of machines")
plt.title("tempature distribution")
plt.show()
print("-----------------------------------------------------------")
plt.figure(figsize=(7,5))
plt.scatter(df["tempature"],df["vibration"])
plt.xlabel("tempature")
plt.ylabel("vibration")
plt.title("tempature vs vibration")
plt.show()
print("-----------------------------------------------------------")
plt.figure(figsize=(8,6))
sns.heatmap(correlation,annot=True,cmap="coolwarm",fmt=".2f")
plt.title("correlation heatmap")
plt.show()
print("-----------------------------------------------------------")
print("temperature of meachine high")
print(df[df["tempature"]>85])
```
