```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("week10 dataset.csv")
print("Original Dataset:")
print(df)
print("\nMissing Values:")
print(df.isnull().sum())
print("\nDuplicate Records:", df.duplicated().sum())
print("\nInvalid Temperature Values:")
print(df[df["tempature"] < 0])
print("\nInvalid Pressure Values:")
print(df[df["pressure"] < 0])
print("\nInvalid Vibration Values:")
print(df[df["vibration"] < 0])
df = df.drop_duplicates()
numeric_columns = [
    "tempature",
    "pressure",
    "vibration",
    "operating-hours"
]
df[numeric_columns] = df[numeric_columns].fillna(
    df[numeric_columns].mean()
)
sensitive_columns = ["Name", "Phone", "Email", "Address"]
df = df.drop(
    columns=[c for c in sensitive_columns if c in df.columns],
    errors="ignore"
)
print("\nMachine Status Distribution:")
print(df["Status"].value_counts())
print("\nResponsible Data Usage Report")
print("--------------------------------")
print("1. Data quality checked")
print("2. Duplicate records removed")
print("3. Missing numerical values handled")
print("4. Invalid sensor values checked")
print("5. Sensitive information removed if present")
print("6. Machine status distribution examined")
print("7. Data should be used only for the intended purpose")
df["status"].value_counts().plot(kind="bar")
plt.title("Machine Status Distribution")
plt.xlabel("Machine Status")
plt.ylabel("Number of Machines")
plt.show()
print("\nEthical analysis completed successfully!")
```
